# Lesson 7 — Imports and Libraries

Python comes with many useful tools, but not everything is available automatically.

Python organizes additional tools into **modules** and **libraries** that your program can import when it needs them.

In this lesson, you'll learn how to:

- understand modules and libraries
- use `import`
- call functions from an imported module
- use `from ... import ...`
- understand why programs use imports
- recognize the imports you'll soon use with your Betabox car

You won't need your car for this lesson.

## 1. Python Has More Tools

You've already used functions such as:

    print()
    len()
    type()
    abs()
    round()

These are **built-in functions**. Python makes them available automatically.

Python has many more capabilities that aren't loaded into every program automatically.

Instead, related tools are organized into **modules**.

A **module** is Python code that provides useful functions, values, or other tools.

For example, Python includes a module named:

    math

It contains tools for working with mathematics.

Before we use those tools, we import the module.

## 2. Importing a Module

The `import` keyword makes a module available to your program.

Run this cell:

```python
import math
```

Nothing appeared underneath the cell, but Python loaded the `math` module and made its tools available.

We can now use them.

```python
print(math.sqrt(16))
```

The `math` module contains a function named:

    sqrt()

which calculates a square root.

We access it using:

    math.sqrt()

Read this as:

> Use the `sqrt()` function from the `math` module.

The `.` connects the module to something it provides.

## 3. One Module Can Provide Many Tools

Importing a module gives your program access to the tools that module provides.

The `math` module contains many mathematical tools.

```python
import math

print(math.sqrt(25))
print(math.floor(4.8))
print(math.ceil(4.2))
```

These functions perform different jobs:

- `math.sqrt()` calculates a square root
- `math.floor()` rounds down
- `math.ceil()` rounds up

They all come from the same `math` module.

You don't need to memorize these functions. The important idea is how we accessed them.

## 4. Modules Can Provide Values Too

Modules don't only contain functions.

They can provide useful values as well.

For example:

```python
import math

print(math.pi)
```

`math.pi` contains an approximation of the mathematical value π.

Notice the difference:

    math.sqrt(16)

uses parentheses because `sqrt` is a function we are calling.

    math.pi

doesn't use parentheses because `pi` is a value.

This is another reason it's useful to understand what a module provides.

## 5. Where Do Imports Go?

Imports are usually placed near the **top of a Python program**.

For example:

```python
import math

number = 36
answer = math.sqrt(number)

print(answer)
```

Putting imports near the top makes it easy to see which modules a program needs.

As your programs become larger, you'll often see several imports together at the beginning.

## 6. Import Once, Use Many Times

You only need to import a module once during a Python session.

After running:

```python
import math
```

you can use `math` in later cells:

This works because the notebook's kernel remembers that you imported `math`.

But remember what you learned in Lesson 1:

> Restarting the kernel clears what Python remembers.

If you restart the kernel and run only the second cell, Python won't know what `math` is yet.

You would need to run the import again.

This is one reason imports are normally placed near the beginning of a notebook or program.

## 8. Importing One Thing From a Module

Sometimes a program only needs a particular tool from a module.

Python lets us import that tool directly.

Instead of:

```python
import math

print(math.sqrt(16))
```

we can write:

```python
from math import sqrt

print(sqrt(16))
```

This instruction:

    from math import sqrt

means:

> From the `math` module, import `sqrt`.

Because `sqrt` was imported directly, we can call:

    sqrt(16)

instead of:

    math.sqrt(16)

## 9. Two Import Styles

You've now seen two common ways to import Python tools.

### Import the module

    import math

Then use:

    math.sqrt(16)

### Import something from the module

    from math import sqrt

Then use:

    sqrt(16)

Both are valid.

Which style a program uses depends on what it needs and how the code is organized.

## 10. Modules and Libraries

You'll hear several words used when talking about reusable Python code.

A **module** is a collection of Python code that provides things your program can use.

A **library** is a larger collection of reusable code and may contain many modules and other components.

Python itself includes a large collection of modules.

Other libraries can be installed to give Python completely new capabilities.

For example, a library could provide tools for:

- creating graphics
- working with data
- communicating over a network
- using a camera
- controlling hardware
- programming a robot

This is how Python can be used for much more than the built-in features you've seen so far.
