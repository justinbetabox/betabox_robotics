## 1. Comparing Values

A **comparison** asks a question about two values.

The result of a comparison is always one of the Boolean values you learned about in the last lesson:

    True
    False

Let's try one.

```python
print(10 > 5)
```

Python compared the two numbers and asked:

> Is `10` greater than `5`?

The answer is `True`.

Before running each comparison below, predict whether Python will produce `True` or `False`.

```python
print(3 > 8)
print(7 < 10)
print(5 == 5)
print(4 == 9)
```

Python has several **comparison operators**:

- `==` equal to
- `!=` not equal to
- `<` less than
- `>` greater than
- `<=` less than or equal to
- `>=` greater than or equal to

Notice that checking whether two values are equal uses:

    ==

A single `=` means **assignment**:

    speed = 50

Two equal signs mean **comparison**:

    speed == 50

## 3. Making a Decision With `if`

A comparison tells us whether something is `True` or `False`.

An **`if` statement** lets Python run code only when a condition is `True`.

Run this example:

```python
distance = 10

if distance < 20:
    print("Object is close!")
```

Read the program like this:

> **If** `distance` is less than `20`, display `"Object is close!"`.

The condition:

    distance < 20

is `True`, so Python runs the indented instruction.

Now look at the next example.

```python
distance = 30

if distance < 20:
    print("Object is close!")
```

This time nothing was printed.

That's not an error.

The condition was `False`, so Python skipped the indented instruction.

## 4. Indentation Matters

Notice the spaces before `print()`:

    if distance < 20:
        print("Object is close!")

Those spaces are called **indentation**.

Indentation tells Python which instructions belong to the `if` statement.

Python uses indentation as part of its syntax, so it matters how your code is lined up.

```python
distance = 10

if distance < 20:
    print("Object is close!")
    print("Be careful!")

print("Distance check complete.")
```

The first two `print()` instructions are indented, so they only run when the condition is `True`.

The final instruction is not indented:

    print("Distance check complete.")

It runs whether the condition is `True` or `False`.

## 5. Choosing Between Two Actions

Sometimes we want one action when a condition is `True` and a different action when it is `False`.

Python uses `else` for this.

```python
distance = 25

if distance < 20:
    print("Object is close!")
else:
    print("Path is clear.")
```

Python checks the condition:

    distance < 20

If it is `True`, Python runs the code under `if`.

Otherwise, Python runs the code under `else`.

Only one of these two branches will run.

## 6. More Than Two Choices

Programs sometimes need more than two possible paths.

Python uses `elif`, which means **else if**.

Imagine we want to classify a battery reading.

```python
battery_voltage = 7.5

if battery_voltage < 7.0:
    print("Battery critical!")
elif battery_voltage < 7.4:
    print("Battery low.")
else:
    print("Battery okay.")
```

Python checks the conditions from **top to bottom**.

As soon as it finds a condition that is `True`, it runs that branch and skips the others.

Try these values one at a time:

    6.8
    7.2
    7.6

Predict the message before running each test.

### Order Matters

Look carefully at the conditions:

    if battery_voltage < 7.0:
        ...
    elif battery_voltage < 7.4:
        ...

Python checks the most urgent condition first.

A value of `6.8` is less than both `7.0` and `7.4`, but Python stops after the first matching branch.

When using `if` and `elif`, the order of your conditions can matter.

## 8. Combining Conditions

Sometimes one comparison isn't enough.

Python can combine conditions using:

- `and`
- `or`
- `not`

Let's start with `and`.

```python
battery_voltage = 7.8
car_ready = True

if battery_voltage >= 7.4 and car_ready:
    print("Ready to start!")
```

With `and`, **both conditions** must be `True`.

In this example:

    battery_voltage >= 7.4

and:

    car_ready

must both be true before the message is displayed.

### Using `or`

With `or`, at least one condition must be `True`.

```python
object_close = True
battery_low = False

if object_close or battery_low:
    print("Car should stop.")
```

The message appears because at least one condition is `True`.

### Using `not`

`not` reverses a Boolean value.

`True` becomes `False`, and `False` becomes `True`.

```python
car_ready = False

if not car_ready:
    print("Car is not ready.")
```

You don't need to build complicated conditions yet.

For now, remember:

- `and` — both conditions must be true
- `or` — at least one condition must be true
- `not` — reverses true and false
