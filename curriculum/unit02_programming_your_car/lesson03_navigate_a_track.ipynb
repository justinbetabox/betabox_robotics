# Lesson 3 — Navigate a Track

In Lesson 2, you learned how to control your Betabox car's motors and steering.

You practiced moving forward and backward, changing speed, steering left and right, and stopping safely.

Now it's time to put those skills to work!

## Your Mission

Program your Betabox car to navigate a marked course using Python.

You'll need to combine the commands you've already learned to control:

- **Speed:** How quickly the car moves.
- **Direction:** Whether the car moves forward or backward.
- **Steering:** Which direction the front wheels point.
- **Duration:** How long each movement lasts.

Unlike the previous lesson, your car will now move across the ground.

You'll experiment with different values, observe the results, and improve your program until your car completes the course.

### What You'll Build

By the end of this lesson, you'll have a Python program that guides your car through a simple track.

Your program should:

1. Start from a designated position.
2. Follow a route containing straight sections and turns.
3. Reach the finish area.
4. Stop safely when the route is complete.

You won't need any new Python concepts or hardware commands.

Instead, you'll demonstrate that you can use what you've already learned to solve a real robotics challenge.

**Remember:** Your teacher must approve the driving area before you begin.

## 1. Your Turn — Drive a Curved Path

Now it's time to see how steering changes the path your car follows.

So far, you've tested the motors with the car supported off the ground.

For this exercise, you'll place the car on the ground and let it move.

**Only begin this exercise when your teacher has approved your driving area.**

### Prepare Your Driving Area

Before you begin:

1. Choose a flat, open area with plenty of space.
2. Remove obstacles from the car's path.
3. Make sure the car cannot drive off a table or raised surface.
4. Keep your hands and feet clear of the wheels.
5. Make sure you can reach the car's power switch if necessary.

### Your Challenge

Write a program that:

1. Centers the steering.
2. Turns the front wheels left by 15 degrees.
3. Moves forward at speed `20`.
4. Continues moving for `0.5` seconds.
5. Stops the drive motors.
6. Returns the steering to the center.

Use `try` and `finally` to help ensure the motors stop and the steering returns to center.

### Test Your Program

Run your program and observe the path your car follows.

If the car doesn't have enough room to move safely, stop the experiment and ask your teacher for help.

Once your first test works, try changing the steering angle to `25` degrees.

Keep the speed and duration the same.

### Think About It

- How did changing the steering angle affect the path?
- Did the car travel farther when you increased the steering angle?
- What would happen if you changed `car.left(15)` to `car.right(15)`?
- Why is it important to stop the motors before starting another test?

**Remember:** Only change one value at a time when experimenting. This makes it easier to understand what caused the car's behavior to change.

## 2. Project — Navigate a Track

Now it's time to put your driving and steering skills to the test!

Your challenge is to program your Betabox car to navigate a marked course using the commands you learned in Lesson 2.

Unlike the previous exercises, you won't be given the exact movement commands.

You'll need to plan your route, test your program, and make adjustments until your car reaches the finish area.

### Set Up Your Track

Your teacher will provide a simple course marked on the floor.

The course should include:

- a clearly marked starting position
- at least one straight section
- at least one left or right turn
- a clearly marked finish area

Make sure the course is on a flat surface with plenty of open space around it.

**Important:** Only begin driving when your teacher has approved the course and testing area.

### Plan Your Route

Before writing your program, look at the course and think about how your car will need to move.

Consider:

1. Where should the car start?
2. Which direction should it face?
3. When should it move straight?
4. When should it turn?
5. How long should each movement last?
6. How will you make sure the car stops at the finish?

You may want to sketch the route or write down the sequence of movements before coding.

### Build Your Program

Write a Python program that:

1. Imports `time` and `BetaboxCar`.
2. Creates a new car object.
3. Uses movement and steering commands to navigate the course.
4. Uses `time.sleep()` to control movement duration.
5. Stops the motors between movement stages when appropriate.
6. Stops the car, centers the steering, and closes the car when finished.

Use `try` and `finally` to help ensure your program stops the motors and releases control.

### Test Your Program

First, test your movement sequence with the car securely supported and its driven wheels off the ground.

Once the commands work correctly, ask your teacher for permission to test on the ground.

Start the car from the marked starting position.

Run your program and observe where the car travels.

It may not reach the finish area on your first attempt.

That's okay! Robotics involves testing and improving your programs.

### Improve Your Program

If your car doesn't follow the intended route, change one value at a time.

You might need to adjust:

- **Speed:** How quickly the motors turn.
- **Steering angle:** How sharply the car turns.
- **Duration:** How long the car moves before the next instruction.

After each change, test your program again.

Try to understand why each adjustment changes the car's path.

### Demonstrate Mastery

Your project is complete when:

- Your car starts from the designated position.
- It follows the intended route through the course.
- It reaches the finish area.
- It stops safely at the end.
- You can explain how your code controls the car's movement.

### Extra Challenge

Once your car completes the course, try one of these extensions:

- Navigate the course in the opposite direction.
- Complete the course using a different speed.
- Create a reusable function for one of your movement sequences.

### Think About It

- Which part of the course was the most difficult?
- How did changing steering angles affect your turns?
- How did changing movement duration affect the distance traveled?
- Why might the same program behave differently on different surfaces?
- What would you change if you had to navigate a more complicated course?

**Remember:** Your goal isn't just to reach the finish. It's to understand how your Python program controls the car.

# Lesson Complete!

Congratulations! You've completed your first Betabox driving challenge!

You used Python to control your car's motors and steering and guide it through a physical course.

You practiced how to:

- plan a sequence of movements
- combine forward movement and steering
- control speed and movement duration
- test your program on a real track
- adjust steering angles and timing to improve your results
- stop your car safely
- solve problems through testing and experimentation

You've also learned an important robotics lesson:

**A program that works in theory might need adjustments to work in the real world.**

Even when you give your car the same instructions, factors such as battery voltage, surface friction, and starting position can affect how it moves.

By testing your program, observing the results, and making changes, you've practiced the same process engineers use to improve their designs.
