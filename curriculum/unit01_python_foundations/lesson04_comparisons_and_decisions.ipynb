## Try it 4: Choose between three actions

These pretend battery voltages use example thresholds, not real battery safety settings.

Now swap the two conditions and their messages. What happens to a reading of 6.8? Restore the original order before moving on.

```python
battery_voltage = 7.5

if battery_voltage < 7.0:
    print("Battery critical!")
elif battery_voltage < 7.4:
    print("Battery low.")
else:
    print("Battery okay.")
```
