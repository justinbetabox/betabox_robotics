## 1. How Ultrasonic Sensors Work

Your Betabox car has an ultrasonic sensor mounted near the front.

An ultrasonic sensor measures distance using sound waves that humans cannot normally hear.

It works in three steps:

1. **Send:** The sensor sends out a short ultrasonic sound pulse.
2. **Reflect:** The sound travels through the air and bounces off an object.
3. **Receive:** The sensor detects the returning echo.

By measuring how long the sound takes to travel to the object and back, the car can estimate how far away the object is.

### Measuring Distance

The Betabox Robotics library reports ultrasonic distance measurements in **centimeters (cm)**.

For example:

- `10 cm` means an object is relatively close.
- `30 cm` means the object is farther away.
- `100 cm` means the object is one meter away.

A smaller measurement generally means an object is closer to the sensor.

### What Can Affect a Measurement?

Ultrasonic sensors aren't perfect.

Measurements can be affected by:

- objects that are too close or too far away
- surfaces that absorb sound
- angled surfaces that reflect sound away from the sensor
- objects that aren't directly in front of the sensor
- movement while the measurement is being taken

This means your program should never assume every sensor measurement is accurate.

## 4. Creating Multiple Warning Levels

An `if` statement allows Python to respond when a condition is true.

But what if we want our program to recognize several different situations?

For example, we might want to display:

- `"Very close!"` when an object is less than 10 centimeters away.
- `"Close"` when an object is between 10 and 20 centimeters away.
- `"Clear"` when an object is 20 centimeters away or farther.

We can use `if`, `elif`, and `else` to make these decisions.

### Checking Multiple Conditions

Look at this example:

```python
if distance < 10:
    print("Very close!")
elif distance < 20:
    print("Close")
else:
    print("Clear")
```

Python checks the conditions from top to bottom.

1. First, it checks whether the distance is less than 10.
2. If that condition is false, it checks whether the distance is less than 20.
3. If neither condition is true, it runs the instructions under `else`.

Only one message will appear.

### Try It With Your Car

Run the next cell:

````python
distance = car.distance()

print("Distance:", distance, "cm")

if distance < 10:
    print("Very close!")
elif distance < 20:
    print("Close")
else:
    print("Clear")
````

### What Happened?

Your program used the distance measurement to choose one of three messages.

For example:

| Distance | Message |
|---|---|
| 5 cm | Very close! |
| 15 cm | Close |
| 30 cm | Clear |

### Why Does the Order Matter?

Python checks the conditions in order.

Imagine checking whether the distance is less than 20 before checking whether it's less than 10.

An object 5 centimeters away would satisfy the first condition.

Python would display `"Close"` and skip the remaining conditions.

That's why we check the smaller distance first.

## 5. Understanding Sensor Limitations

Your Betabox car uses an ultrasonic sensor to measure the distance to objects in front of it.

Although this sensor is useful, its measurements aren't always perfect.

### Why Measurements Can Be Inaccurate

Several things can affect an ultrasonic distance measurement:

- **Object position:** An object might not be directly in front of the sensor.
- **Object shape:** Curved or angled surfaces can reflect sound away from the sensor.
- **Object material:** Some materials absorb sound instead of reflecting it clearly.
- **Object distance:** Objects that are extremely close or far away may be difficult to measure.
- **Movement:** Measurements can change while the car or an object is moving.

Sometimes, the sensor might return an unexpected measurement or fail to detect an object.

### What Happens When a Measurement Is Invalid?

Imagine your program receives a distance measurement that is missing or doesn't make sense.

If your program assumes the path is clear, the robot might make an unsafe decision.

Instead, we should check whether the measurement is valid before using it.

For example, we can use:

```python
if distance is None or distance <= 0:
    print("Sensor error!")
```

This checks whether the measurement is missing or is zero or negative.

If either condition is true, Python displays a warning.

We can also check for values that aren't finite numbers.

### Checking Sensor Measurements

Let's write a program that checks whether the distance measurement is usable.

We'll use Python's `math.isfinite()` function, which checks whether a number is finite.

Run the next cell.

````python
import math

distance = car.distance()

if (
    distance is None
    or not math.isfinite(distance)
    or distance <= 0
):
    print("Sensor error!")
else:
    print("Distance:", distance, "cm")
````

### What Happened?

Your program checked the sensor measurement before displaying it.

If the measurement was valid, Python displayed the distance in centimeters.

Otherwise, it displayed:

```text
Sensor error!
```

### Why Does This Matter?

In the previous sections, we used the sensor measurement directly in our conditions.

For example:

```python
if distance < 20:
    print("Obstacle detected!")
```

That works when the measurement is valid.

However, a missing or invalid measurement could cause an error or lead to an incorrect decision.

By checking the measurement first, we can make our programs more reliable.

**Remember:** A valid-looking measurement isn't guaranteed to be accurate. It simply means the value passed our basic checks.
