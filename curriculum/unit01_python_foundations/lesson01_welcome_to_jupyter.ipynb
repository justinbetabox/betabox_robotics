## Try it 3: Interrupt a running cell

The next cell normally finishes instantly. To practice stopping a program:

1. Change `wait_seconds = 0` to `wait_seconds = 15`.
2. Run the cell, then choose **Kernel → Interrupt Kernel** while it's waiting.
3. You may see `KeyboardInterrupt`. That's an expected result of stopping the cell.
4. **Change `wait_seconds` back to `0` and run the cell once more** so the notebook can run quickly later.

```python
import time

wait_seconds = 0  # Change to 15 for the interrupt exercise.
time.sleep(wait_seconds)
print("Wait complete!")
```
