## 1. Giving Python Instructions

A Python program is made of **instructions**.

Python reads those instructions and carries them out.

You've already used one Python instruction:

```python
print("Hello, Betabox!")
```

`print()` tells Python to display something.

The information inside the parentheses tells `print()` what to display.

Try running the next few examples.

```python
print("Python is running!")
print(42)
print(2 + 3)
```

Notice that Python ran the instructions **from top to bottom**.

Each `print()` instruction displayed a result.

## 2. Python Can Do Math

Python can work with numbers and perform calculations.

Before running each example, try to predict the result.

```python
print(5 + 3)
print(10 - 4)
print(6 * 2)
print(20 / 4)
```

Python uses symbols called **operators** to perform calculations:

- `+` adds
- `-` subtracts
- `*` multiplies
- `/` divides

These symbols will become useful later when your robot works with speeds, distances, sensor readings, and other numbers.

### More Than One Operation

Python can perform more complicated calculations too.

Predict the result before running this cell:

```python
print(2 + 3 * 4)
```

Did you predict `20` or `14`?

Python follows an **order of operations**, just like mathematics. Multiplication happens before addition.

You can use parentheses when you want part of a calculation to happen first:

```python
print((2 + 3) * 4)
```

Now the result is `20`.

You don't need to memorize every rule today. Just remember that parentheses can make calculations clearer and control which part happens first.

## 3. Working With Text

Python calls a piece of text a **string**.

Strings are surrounded by quotation marks:

print("Hello!")
print("Betabox")
print("I am learning Python.")The quotation marks tell Python that the characters inside them are text.

You can use either double quotation marks:

    "Hello!"

or single quotation marks:

    'Hello!'

For now, we'll usually use double quotation marks in our examples.

### Combining Text

Python can join strings together using `+`.

    Betabox

## 4. Comments

Programs can contain notes for the people reading the code.

These notes are called **comments**.

A Python comment begins with `#`.

Python ignores everything after the `#` on that line.

```python
# This is a comment.
print("Python runs this line.")

# Python ignores this line too.
print("And this one runs!")
```

Comments can explain what code does or remind you why you wrote it.

Later, comments will help make your robot programs easier to understand.

## 5. Mistakes Are Part of Programming

Python has rules about how code must be written.

When Python can't understand an instruction, it displays an **error**.

Errors are normal. Every programmer sees them.

Let's create one on purpose.

```python
print("Hello, Betabox!"
```

Run the cell above.

You should see an error instead of our message.

Look at the code carefully.

The opening `(` has a matching `)`, and the opening `"` has a matching `"`.

When symbols such as quotation marks or parentheses don't match correctly, Python may not be able to understand the instruction.

## 6. Python Is Case-Sensitive

Python treats uppercase and lowercase letters as different characters.

For example:

    print

and:

    Print

are not the same instruction.

Try the next cell.

When writing Python, spelling, capitalization, punctuation, and indentation can all matter.
