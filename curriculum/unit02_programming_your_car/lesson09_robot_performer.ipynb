## 4. Combining Movement and Audio

You've tested your robot's movement sequence and experimented with speech and music.

Now it's time to bring everything together!

A robot performance combines different actions into one organized program.

The order and timing of those actions determine how the performance looks and sounds.

### Coordinating Your Actions

Imagine your robot performing this sequence:

1. Introduce itself using speech.
2. Play a short greeting melody.
3. Move forward.
4. Stop and play a musical note.
5. Steer left and follow a curved path.
6. Stop and center its steering.
7. Play a final melody.
8. Say goodbye.

Each action contributes to the overall performance.

### Understanding Timing

Movement commands and audio commands don't always behave the same way.

For example:

```python
car.forward(20)
time.sleep(1)
car.stop()
```

This moves the car for approximately one second.

However, an audio command might continue playing while Python moves to the next instruction.

You can check whether audio is still playing using:

```python
car.is_audio_playing()
```

If you want to wait for audio playback to finish, you can use a loop:

```python
while car.is_audio_playing():
    time.sleep(0.1)
```

This checks whether audio is still playing and waits until playback finishes.

If you use this technique, consider adding a time limit so your program cannot wait forever if audio playback becomes stuck.

### Keeping Movements Safe

Your robot should never continue driving simply because it's waiting for an audio command.

A safer approach is to:

1. Stop the drive motors.
2. Play the desired sound.
3. Wait for the sound to finish if necessary.
4. Begin the next movement.

This makes the performance easier to test and helps prevent unexpected movement.

### Organizing Your Program

Your program should have a clear structure:

**Introduction**

- Speak a greeting.
- Play a short melody.

**Performance**

- Complete your planned movement sequence.
- Stop between movement stages.
- Play sounds or speech at selected moments.

**Finale**

- Stop the motors.
- Center the steering.
- Play a final sound.
- Say goodbye.
