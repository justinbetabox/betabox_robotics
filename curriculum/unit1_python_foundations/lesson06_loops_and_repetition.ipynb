# Lesson 6 — Loops and Repetition

Programs often need to repeat instructions.

You might want a program to:

- display a message several times
- count through a sequence of numbers
- repeat an action
- keep checking a condition until something changes

Python uses **loops** to repeat code.

In this lesson, you'll learn how to:

- repeat instructions with `for`
- use `range()`
- work with a loop variable
- repeat code with `while`
- stop a loop with `break`
- avoid accidental infinite loops

You won't need your car for this lesson.

## 1. Repeating Instructions

Suppose you wanted Python to display `"Hello!"` three times.

You could write:

```python
print("Hello!")
print("Hello!")
print("Hello!")
```

That works, but repeating the same instruction over and over makes programs longer and harder to change.

A **loop** lets us tell Python to repeat code for us.

## 2. Your First `for` Loop

A `for` loop is useful when you know how many times you want something to repeat.

Run this example:

```python
for number in range(3):
    print("Hello!")
```

The indented instruction ran three times.

Read the first line like this:

> For each number in a sequence of 3 numbers, run the indented code.

Just like an `if` statement, the code that belongs to the loop is **indented**.

## 3. Using `range()`

The `range()` function creates a sequence of numbers.

Run this cell:

```python
for number in range(5):
    print(number)
```

You should see:

    0
    1
    2
    3
    4

`range(5)` creates five values, beginning with `0`.

This can feel unusual at first.

A useful rule is:

> `range(5)` repeats the loop five times.

### Why Does Python Start at 0?

You may be wondering why:

    range(5)

produces:

    0
    1
    2
    3
    4

instead of:

    1
    2
    3
    4
    5

Python often starts counting **positions** at `0`.

You can think of `0` as meaning:

> Start at the beginning — move zero positions forward.

Then:

- `0` is the first position
- `1` is the second position
- `2` is the third position
- `3` is the fourth position

This way of counting is called **zero-based indexing**.

You'll see it again when you learn how Python stores groups of values.

Imagine four boxes in a row:

    Position:    0       1       2       3
               ┌───┐   ┌───┐   ┌───┐   ┌───┐
               │ A │   │ B │   │ C │   │ D │
               └───┘   └───┘   └───┘   └───┘

The first box is at position `0`, not position `1`.

This may feel strange at first, but it will become familiar as you use Python.

## 4. The Loop Variable

In:

    for number in range(5):

`number` is a variable.

Each time the loop repeats, Python gives it the next value from `range()`.

```python
for number in range(5):
    print("Loop number:", number)
```

The value stored in `number` changes automatically during the loop.

## 5. Repeating Multiple Instructions

A loop can repeat more than one instruction.

All code indented at the same level belongs to the loop.

```python
for number in range(3):
    print("Starting step", number)
    print("Step complete")

print("Loop finished")
```

The first two `print()` instructions run three times.

The final instruction is not indented, so it runs once after the loop finishes.

## 6. Loops Can Change Variables

Loops can update variables each time they repeat.

Run this example:

```python
total = 0

for number in range(5):
    total = total + 1

print(total)
```

The loop runs five times.

Each time, it adds `1` to `total`.

At the end:

    total = 5

## 7. Using the Loop Value

The loop variable can be used in calculations too.

```python
for number in range(5):
    doubled = number * 2
    print(doubled)
```

Each time through the loop:

1. `number` gets a new value
2. Python doubles it
3. the result is displayed

This lets one small block of code work with many values.

## 8. Choosing a Start and Stop

`range()` can also be given a starting value.

Run:

```python
for number in range(1, 6):
    print(number)
```

This displays:

    1
    2
    3
    4
    5

When `range()` has two arguments:

    range(start, stop)

Python begins at `start` and stops **before** `stop`.

So:

    range(1, 6)

means:

> Start at `1` and stop before `6`.

## 9. Repeating While Something Is True

A `for` loop is useful when you know how many times to repeat.

A **`while` loop** repeats while a condition remains `True`.

Run this example:

```python
count = 0

while count < 3:
    print("Count:", count)
    count = count + 1
```

Python checks:

    count < 3

before each repetition.

As long as the condition is `True`, the loop continues.

Inside the loop, `count` increases.

Eventually:

    count < 3

becomes `False`, and the loop stops.

## 10. Predict Before You Run

What values will this program display?

Make your prediction before running it.

```python
number = 5

while number > 0:
    print(number)
    number = number - 1
```

This kind of loop is often called a **countdown**.

The variable changes each time until the condition becomes `False`.

## 11. Be Careful With `while`

A `while` loop can keep running forever if its condition never becomes `False`.

For example:

    while True:
        print("Running")

`True` never changes, so the loop never ends on its own.

Loops like this are called **infinite loops**.

Infinite loops are sometimes useful, especially in robotics, but you need to know how to stop them safely.

### Try an Infinite Loop Safely

Run the next cell.

Let it run a few times, then use Jupyter's **Interrupt Kernel** command to stop it.

```python
import time

while True:
    print("Running...")
    time.sleep(1)
```

You may see:

    KeyboardInterrupt

That's expected.

You stopped the loop yourself.

Later, car programs may use loops that continuously read sensors or wait for events, so knowing how to interrupt running code is an important safety skill.

## 12. Stopping a Loop With `break`

Python also has a keyword called `break`.

`break` immediately exits the loop.

```python
count = 0

while True:
    print(count)

    if count == 3:
        break

    count = count + 1

print("Loop stopped")
```

The loop itself says:

    while True:

so it could continue forever.

But once:

    count == 3

becomes `True`, Python runs:

    break

and leaves the loop.

## 13. Loops and Decisions Work Together

Loops become especially useful when combined with decisions.

Look at this program:

```python
for number in range(5):
    if number == 3:
        print("Found 3!")
    else:
        print("Current number:", number)
```

Each time the loop repeats, the `if` statement checks the current value.

This pattern will become very useful in car programs:

1. repeat
2. check something
3. decide what to do

## 14. Thinking Like a Computer Program

Imagine these numbers came from a distance sensor:

    50
    35
    18
    10

We can use a loop to check them one at a time.

```python
distances = [50, 35, 18, 10]

for distance in distances:
    print("Distance:", distance)

    if distance < 20:
        print("Obstacle detected!")
```

Don't worry about the square brackets yet.

For now, think of `distances` as a group of values.

The loop takes each value, stores it in `distance`, and then checks whether the object is close.

This is similar to the kind of logic you'll soon use with a real car sensor.
