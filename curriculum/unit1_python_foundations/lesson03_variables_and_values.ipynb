## 1. Creating a Variable

A **variable** gives a name to a value.

Run the next cell:

```python
car_name = "Betabox"
```

Nothing appeared underneath the cell, but Python did something important.

It stored the text `"Betabox"` and gave it the name:

    car_name

We can now use that name somewhere else in our program.

```python
print(car_name)
```

Python looked up the value stored in `car_name` and displayed it.

The `=` symbol is called the **assignment operator**.

This instruction:

    car_name = "Betabox"

means:

> Store the value `"Betabox"` in the variable named `car_name`.

## 2. Variables Can Store Different Kinds of Values

Variables aren't limited to text.

They can store numbers and other kinds of information too.

Run the next cell.

```python
car_name = "Betabox"
speed = 50
distance = 25.5
ready = True

print(car_name)
print(speed)
print(distance)
print(ready)
```

These variables contain different kinds of values:

- `"Betabox"` is a **string** (`str`) — text
- `50` is an **integer** (`int`) — a whole number
- `25.5` is a **float** (`float`) — a number with a decimal
- `True` is a **Boolean** (`bool`) — either `True` or `False`

You'll see all four kinds of values while programming your car.

### Check the Type

Python's `type()` function can tell you what kind of value something is.

Run the next cell.

```python
print(type(car_name))
print(type(speed))
print(type(distance))
print(type(ready))
```

You should see results containing:

    str
    int
    float
    bool

You don't need to memorize everything about these types yet.

For now, remember that Python can work with different kinds of values.

## 3. Variables Can Change

The value stored in a variable doesn't have to stay the same.

Run these instructions from top to bottom.

```python
speed = 25
print(speed)

speed = 50
print(speed)

speed = 75
print(speed)
```

The variable `speed` was assigned three different values.

Each new assignment replaced the value that was stored before it.

After the final instruction, `speed` contains:

    75

## 4. Using Variables in Calculations

Variables can be used anywhere Python expects a value.

That means we can use them in calculations.

```python
length = 5
width = 3

print(length * width)
```

Python used the values stored in `length` and `width` to perform the calculation.

We can also store the result in another variable:

```python
length = 5
width = 3

area = length * width

print(area)
```

Read this program from top to bottom:

1. `length` stores `5`
2. `width` stores `3`
3. Python multiplies `length` by `width`
4. the result is stored in `area`
5. `print(area)` displays the result

Breaking a program into named values can make it much easier to understand.

## 5. Updating a Value

A program can use a variable's current value to calculate its next value.

Run this cell:

```python
score = 10

score = score + 5

print(score)
```

At first:

    score = 10

Then Python evaluates:

    score + 5

which produces `15`.

Finally, that new value is stored back in `score`.

So:

    score = score + 5

means:

> Take the current value of `score`, add `5`, and store the new value back in `score`.

This may look unusual if you've seen `=` in mathematics.

In Python, `=` performs **assignment**.

## 6. Displaying Variables in Messages

Sometimes we want to display text and a variable together.

One simple way is to give `print()` more than one value:

```python
car_name = "Betabox"

print("Car name:", car_name)
```

`print()` displays both values and adds a space between them.

This works with numbers too:

```python
speed = 50

print("Speed:", speed)
```

## 7. Choosing Variable Names

Good variable names describe the information they contain.

Compare:

    x = 50

with:

    speed = 50

Both work, but `speed` tells us what the value means.

Python variable names:

- can contain letters, numbers, and underscores
- cannot begin with a number
- cannot contain spaces
- are case-sensitive

We will usually use lowercase words separated by underscores:

    car_name
    sensor_distance
    battery_voltage

This style is called **snake_case**.

### Which Name Is Better?

Which variable name would make a program easier to understand?

## 8. True and False

A **Boolean** stores one of two values:

    True
    False

Booleans are useful when a program needs to remember whether something is on, off, ready, detected, connected, or available.

```python
robot_ready = True
obstacle_detected = False

print(robot_ready)
print(obstacle_detected)
```

Notice that `True` and `False` do **not** use quotation marks.

These:

    True
    False

are Boolean values.

These:

    "True"
    "False"

are strings containing text.

We'll use Boolean values to make decisions in the next lesson.
