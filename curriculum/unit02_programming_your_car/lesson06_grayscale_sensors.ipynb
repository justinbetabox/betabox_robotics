```python
from betabox_robotics import BetaboxCar


def read_raw():
    car = BetaboxCar()
    try:
        car.stop()
        return car.line_values()
    finally:
        car.close()


def read_calibrated(threshold=0.5):
    car = BetaboxCar()
    try:
        car.stop()
        reading = car.line_reading(threshold=threshold)
        if reading.normalized is None:
            print("Grayscale calibration is missing. Use the Betabox Calibration tool.")
            return None
        return reading
    finally:
        car.close()


def show_raw(readings):
    left, center, right = readings
    print("Left:", left, "Center:", center, "Right:", right)
```

## 3. Create a raw threshold

Start with the center sensor. Run this cell over the light surface, then over the dark material. Does your classification match the actual surface?

If the two reference measurements are identical or very close, collect better references before continuing.

```python
white_value = floor_values[1]
black_value = line_values[1]

if white_value == black_value:
    print("References are identical. Repeat the reference measurements.")
else:
    threshold = (white_value + black_value) / 2
    center = read_raw()[1]
    print("Center reading:", center, "Raw threshold:", threshold)

    if black_value < white_value:
        dark_detected = center < threshold
    else:
        dark_detected = center > threshold

    if dark_detected:
        print("Dark surface detected!")
    else:
        print("Light surface detected!")
```
