## 1. Groups of Values

Imagine that you have three sensor readings.

You could store them in three separate variables:

```python
reading_one = 120
reading_two = 250
reading_three = 180

print(reading_one)
print(reading_two)
print(reading_three)
```

That works, but what if you had 10 readings? Or 100?

Python lets us group related values together in a **list**.

## 2. Creating a List

A **list** stores multiple values together.

Lists use square brackets:

    [ ]

Run this example:

```python
readings = [120, 250, 180]

print(readings)
```

The variable `readings` contains three values:

    120
    250
    180

The values are separated by commas and surrounded by square brackets:

    [120, 250, 180]

Each value in a list is called an **item** or **element**.

## 3. Lists Can Store Different Kinds of Values

Lists aren't limited to numbers.

For example, we can create a list of colors:

```python
colors = ["red", "green", "blue"]

print(colors)
```

Or a list of Boolean values:

```python
sensor_states = [True, False, True]

print(sensor_states)
```

Lists can even contain different types of values together, although related values are often easier to work with when they use the same type.

## 4. Accessing Items in a List

Sometimes we need one particular item from a list.

Python gives every item a position called an **index**.

Remember zero-based indexing from the loops lesson?

Python starts counting positions at `0`.

For this list:

    readings = [120, 250, 180]

the positions are:

    Index:       0     1     2
               ┌─────┬─────┬─────┐
    Value:     │ 120 │ 250 │ 180 │
               └─────┴─────┴─────┘

The first item is at index `0`.

```python
readings = [120, 250, 180]

print(readings[0])
print(readings[1])
print(readings[2])
```

Square brackets after the variable name select an item:

    readings[0]

means:

> Get the item at index `0` from `readings`.

Remember:

- `[0]` — first item
- `[1]` — second item
- `[2]` — third item

This is one of the places where Python's zero-based indexing becomes especially important.

## 6. Indexes Must Exist

Our list:

    colors = ["red", "green", "blue"]

contains three items.

That means its indexes are:

    0
    1
    2

What do you think will happen if we ask for index `3`?

Run the next cell.

```python
colors = ["red", "green", "blue"]

print(colors[3])
```

You should see an error containing:

    IndexError

Python is telling you that the requested list index doesn't exist.

The list has three items, but because indexing begins at `0`, the final index is `2`.

This is a common programming mistake.

Fix the cell so it displays `"blue"`.

## 7. How Many Items Are There?

You've already used the `len()` function with strings.

It also works with lists.

```python
readings = [120, 250, 180]

print(len(readings))
```

`len(readings)` returns:

    3

because the list contains three items.

Notice an important difference:

- the **length** is `3`
- the indexes are `0`, `1`, and `2`

The number of items is not the same as the final index.

## 8. Changing an Item

Lists can be changed after they are created.

We can assign a new value to a particular index:

```python
speeds = [20, 40, 60]

print(speeds)

speeds[1] = 50

print(speeds)
```

Originally:

    speeds = [20, 40, 60]

Index `1` contained:

    40

This instruction:

    speeds[1] = 50

replaced it with `50`.

The list became:

    [20, 50, 60]

## 9. Adding an Item

Lists can grow while a program is running.

The `append()` method adds an item to the end of a list.

```python
readings = [100, 200]

readings.append(300)

print(readings)
```

The list began as:

    [100, 200]

After:

    readings.append(300)

it became:

    [100, 200, 300]

Notice the syntax:

    readings.append(300)

`append()` belongs to the list stored in `readings`.

You'll see this same `object.something()` pattern frequently when you begin working with your car.

## 10. Building a List One Item at a Time

A list can begin empty:

    []

Then a program can add values as it runs.

```python
readings = []

readings.append(100)
readings.append(150)
readings.append(200)

print(readings)
```

This is useful when a program doesn't know all of its values ahead of time.

For example, a program could collect sensor readings and add each new measurement to a list.

## 11. Looping Through a List

Lists work especially well with `for` loops.

Instead of using `range()`, a loop can take each item directly from a list.

```python
colors = ["red", "green", "blue"]

for color in colors:
    print(color)
```

Each time through the loop, Python takes the next item from `colors` and stores it in the variable `color`.

The loop runs once for every item.

Read:

    for color in colors:

as:

> For each `color` in `colors`, run the indented code.

This is often easier than accessing every item individually:

    print(colors[0])
    print(colors[1])
    print(colors[2])

A loop can work even if the list becomes much larger.

## 12. Working With Sensor-Style Data

Remember the example from the loops lesson?

Now you know what this actually means:

```python
distances = [50, 35, 18, 10]

for distance in distances:
    print("Distance:", distance)

    if distance < 20:
        print("Obstacle detected!")
```

`distances` is a list containing four pretend sensor readings.

The loop:

1. takes one reading from the list
2. stores it in `distance`
3. displays the reading
4. checks whether the reading is less than `20`
5. repeats with the next reading

This combines:

- lists
- variables
- loops
- comparisons
- decisions

Soon, the values won't be pretend — they'll come from your car's real sensors.

## 13. Lists and Loops Together

A loop can also add values to a list.

Run this program:

```python
numbers = []

for number in range(5):
    numbers.append(number)

print(numbers)
```

The list started empty.

Each time the loop repeated, the current value of `number` was added to the list.

The final result is:

    [0, 1, 2, 3, 4]

This is another example of several Python ideas working together.

## 15. Tuples

Python has another collection called a **tuple**.

A tuple looks similar to a list, but it usually uses parentheses:

```python
sensor_values = (120, 250, 180)

print(sensor_values)
print(sensor_values[0])
print(sensor_values[1])
print(sensor_values[2])
```

Tuples use the same kind of zero-based indexing:

- `[0]` — first item
- `[1]` — second item
- `[2]` — third item

You can also loop through a tuple:

```python
sensor_values = (120, 250, 180)

for value in sensor_values:
    print(value)
```

The biggest difference you'll need to know for now is:

> Lists are designed to be changed. Tuples cannot have their individual items replaced after they are created.

You'll sometimes receive groups of values as tuples when working with Python libraries.

For now, you mainly need to be able to recognize and read them.

## 16. Lists and Tuples

Both lists and tuples can group values together.

### List

    values = [10, 20, 30]

Uses square brackets and can be changed.

### Tuple

    values = (10, 20, 30)

Usually uses parentheses and cannot have its individual items replaced.

Both can:

- contain multiple values
- be accessed with indexes
- be used with `len()`
- be used in `for` loops

You'll see both while programming in Python.

## 17. Values With Names

Lists and tuples organize values by position.

Sometimes it's more useful to organize values using **names**.

Python provides a collection called a **dictionary** for this.

```python
car_info = {
    "name": "Betabox",
    "speed": 50,
    "ready": True,
}

print(car_info)
```

A dictionary stores pairs of **keys** and **values**.

In this dictionary:

- `"name"` is connected to `"Betabox"`
- `"speed"` is connected to `50`
- `"ready"` is connected to `True`

Instead of selecting information by a numbered position, we can select it by its key.

```python
car_info = {
    "name": "Betabox",
    "speed": 50,
    "ready": True,
}

print(car_info["name"])
print(car_info["speed"])
print(car_info["ready"])
```

You don't need to memorize everything about dictionaries yet.

For now, you should be able to recognize one when you see it.

Remember:

- lists use numbered positions
- tuples also use numbered positions
- dictionaries use keys to identify values

We'll explore dictionaries more when we have a real reason to use them.
