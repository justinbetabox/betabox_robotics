# Lesson 5 — Functions, Arguments, and Return Values

You've already used Python functions like:

    print()
    type()

A **function** is a reusable instruction that performs a task.

Functions can:

- receive information
- perform work
- return a result

In this lesson, you'll learn how to:

- call functions
- give functions arguments
- use more than one argument
- understand return values
- store returned values in variables
- create simple functions of your own

You won't need your car for this lesson.

## 1. Calling a Function

You've already called the `print()` function many times.

Run this cell:

```python
print("Hello!")
```

This is called a **function call**:

    print("Hello!")

The function's name is:

    print

The parentheses tell Python to run the function:

    ()

The information inside the parentheses is given to the function.

## 2. Giving Information to a Function

A value given to a function is called an **argument**.

In:

    print("Hello!")

the string:

    "Hello!"

is an argument.

Try another example:

```python
print(42)
```

This time the argument is the number `42`.

Functions can receive different kinds of arguments depending on what they are designed to do.

### Variables Can Be Arguments Too

An argument doesn't have to be written directly inside the function call.

Python looks up the value stored in `message` and gives that value to `print()`.

Later, you'll do the same thing with car commands.

For example, a speed might be stored in a variable and then passed to a movement function.

## 3. Using More Than One Argument

Some functions can accept more than one argument.

Arguments are separated by commas.

```python
car_name = "Betabox"
speed = 50

print("Car:", car_name)
print("Speed:", speed)
```

In this instruction:

    print("Speed:", speed)

`print()` receives two arguments:

1. `"Speed:"`
2. the value stored in `speed`

## 4. Functions Can Give Information Back

Some functions don't just perform an action.

They can also **return a value**.

You've already used one:

```python
value_type = type(42)

print(value_type)
```

The `type()` function examined the value `42` and **returned** its type.

That returned value was stored in:

    value_type

Then we displayed it with `print()`.

A useful pattern is:

    result = function(argument)

Read this as:

> Run the function using the argument, then store the returned result in `result`.

You'll use this pattern constantly when reading information from your car.

## 5. Functions Don't Always Need Arguments

Some functions can be called without giving them any information.

The parentheses are still required.

```python
print()
```

`print()` can run without an argument. It simply displays a blank line.

Different functions have different rules about:

- how many arguments they need
- which kinds of values they accept
- whether they return a value

Learning to use a function means learning its **interface**.

## 6. Python Has Built-In Functions

Python includes many functions that are ready to use.

Here are a few useful ones.

```python
print(len("Betabox"))
print(abs(-10))
print(round(3.7))
```

These functions do different jobs:

- `len()` returns the length of something
- `abs()` returns the absolute value of a number
- `round()` rounds a number

## 7. Saving a Function's Result

Returned values become especially useful when we store them in variables.

```python
word_length = len("Betabox")

print(word_length)
```

Now `word_length` contains the result returned by `len()`.

We can use that value again later:

```python
word_length = len("Betabox")

double_length = word_length * 2

print(double_length)
```

The value returned by one function can become part of a larger program.

Later, your car might do something like:

1. call a sensor function
2. store the returned reading
3. compare the reading
4. decide what to do

## 8. Functions Can Be Used Inside Expressions

Because a returned value behaves like another value, a function call can be used directly in a calculation.

```python
print(len("car") + 10)
```

Python:

1. calls `len("car")`
2. gets the returned value
3. adds `10`
4. gives the final result to `print()`

You don't always need to store every result in a variable, but variables often make programs easier to read.

## 9. Using Function Results in Decisions

Functions become especially powerful when their results are used in conditions.

```python
name = "Betabox"

if len(name) > 5:
    print("That name has more than 5 characters.")
```

This program combines ideas from several lessons:

- a variable
- a function call
- a returned value
- a comparison
- an `if` statement

This is how larger programs are built: small ideas working together.

## 10. Creating Your Own Function

Python also lets you create functions of your own.

A function definition begins with `def`.

Run this example:

```python
def say_hello():
    print("Hello!")
```

Running that cell **defined** the function, but it did not display anything.

The function now exists and can be called whenever we need it:

```python
say_hello()
```

The indented code belongs to the function.

Every time we call:

    say_hello()

Python runs the instructions inside it.

## 11. Giving Your Function Information

A function you create can receive information too.

The name inside the function definition is called a **parameter**.

```python
def greet(name):
    print("Hello,", name)
```

```python
greet("Alex")
greet("Betabox")
```

In:

    def greet(name):

`name` is a **parameter**.

In:

    greet("Alex")

`"Alex"` is the **argument**.

A parameter is the name used inside the function.

An argument is the actual value supplied when the function is called.

## 12. Returning a Value

A function you create can return information using `return`.

```python
def double(number):
    return number * 2
```

The function received `5`, multiplied it by `2`, and returned `10`.

The returned value was stored in:

    result

## 13. Actions and Results

Functions can be used in different ways.

Some mainly **perform an action**:

    print("Hello!")

Some mainly **return information**:

    length = len("Betabox")

Some functions can do both.

When we begin programming the car, you'll see the same pattern.
