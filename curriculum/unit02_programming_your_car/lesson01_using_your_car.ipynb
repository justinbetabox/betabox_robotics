## 2. Importing `BetaboxCar`

In Lesson 7, you learned how Python programs use imports to access additional capabilities.

Your Betabox includes a Python library named:

    betabox_robotics

From that library, we'll import:

    BetaboxCar

```python
from betabox_robotics import BetaboxCar
```

You already know how to read this:

> From the `betabox_robotics` library, import `BetaboxCar`.

`BetaboxCar` gives Python an interface for working with the Betabox car in front of you.

Running the import doesn't take control of the car yet.

It simply makes `BetaboxCar` available to your program.

## 3. Creating Your Car

Now we're ready to create the Python object that represents your car.

Run the next cell.

```python
car = BetaboxCar()
```

You just connected your Python program to your Betabox car!

This instruction:

    car = BetaboxCar()

does two important things.

First:

    BetaboxCar()

creates a Python **object** that represents your Betabox car.

Then:

    car =

stores that object in a variable named `car`.

From now on, we'll use `car` to interact with your Betabox.

## 4. What Is an Object?

An **object** is a Python value that can represent something and provide ways to interact with it.

In this case:

    car

represents your Betabox car.

The car object provides capabilities for working with things such as:

- movement
- sensors
- audio
- the camera
- vision
- system information

You don't need to understand how all of that works internally.

The `car` object gives your program one convenient way to interact with your Betabox.

## 5. Your Notebook Now Has Control

Something else important happened when you ran:

    car = BetaboxCar()

Your notebook became the active controller of the car.

Betabox allows one application at a time to control hardware such as the motors and servos.

This prevents two programs from trying to control the car at the same time.

While your `car` is active, another application may tell you that the car is busy.

When you're finished, you'll release control by closing the car.

## 6. Objects Have Methods

In Lesson 5, you learned about functions.

Objects can provide their own functions. A function that belongs to an object is called a **method**.

A method is called using a dot:

    object.method()

For your car, that looks like:

    car.method()

You've already seen similar syntax with lists:

    readings.append(100)

`append()` is a method provided by a list.

Now you'll use methods provided by your car.

## 7. Ask Your Car for Information

Let's get some real information from your Betabox.

Your car can measure its battery voltage.

Run:

```python
voltage = car.battery_voltage()

print(voltage)
```

That value came from your real car.

Look closely at:

    car.battery_voltage()

You can read this as:

> Ask `car` for its battery voltage.

The method returns a value, just like the functions you learned about earlier.

We stored that returned value in:

    voltage

Then we displayed it with `print()`.

## 8. Use the Returned Value

Once information from the car is stored in a variable, it works like other Python values.

Run the next cell.

```python
voltage = car.battery_voltage()

print("Battery voltage:", voltage)
```

The important difference is where the value came from.

Earlier, you might have written:

    voltage = 7.5

and chosen the value yourself.

Now:

    voltage = car.battery_voltage()

asks the physical car for a measurement.

Your Python program can now use information from the real world.

## 9. Read the Distance Sensor

Your Betabox car can measure more than just its battery voltage.

It also has an ultrasonic distance sensor that measures how far away an object is.

Unlike the battery voltage, which usually changes slowly, the distance measurement can change immediately when you move something in front of the sensor.

Place your hand or another object in front of the ultrasonic sensor.

Run the next cell to see how far away it is.

```python
distance = car.distance()

print("Distance:", distance, "cm")
```

Try moving the object closer or farther away and running the cell again.

Does the value change?

This time:

    car.distance()

asked a physical sensor for information about the world around your car.

The returned measurement was stored in:

    distance

## 11. Closing Your Car

When you're finished working with your car, you should close it.

Your Betabox can be controlled by different applications, including Jupyter notebooks, Manual Drive, and calibration tools.

However, only one application can control the car's motors and servos at a time.

Closing your car stops its active operations and releases control so another application can use it.

Run the next cell:

```python
car.close()
```

Closing the car safely finishes its active operations and releases control.

This is an important habit.

For now, remember the basic car lifecycle:

    Create
      ↓
    car = BetaboxCar()

    Use
      ↓
    car.something()

    Close
      ↓
    car.close()

After closing the car, don't continue using that `car` object.

Create a new one when you want to take control again.

## 12. Take Control Again

Once you've closed a car object, you cannot use that same object again.

To continue working with your Betabox, create a new car object.

Run the next cell:

```python
from betabox_robotics import BetaboxCar

car = BetaboxCar()
```

If you restart your notebook's kernel, Python forgets the objects and imports created earlier. Run the import and car creation cells again before using the car.

## 13. What If the Car Is Busy?

Sometimes another application may already control the car.

If that happens when you try:

    car = BetaboxCar()

Python may report that car control is already in use.

This is a safety feature.

Don't try to work around it.

Instead:

1. check whether another notebook or Betabox application is using the car
2. stop or close that application
3. try again

Only one application should control the car at a time.

## 14. Methods Can Do Different Jobs
