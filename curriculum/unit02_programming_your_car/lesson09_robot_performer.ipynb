## Get ready

Run the setup cell once. `move_for()` performs a short movement and stops the motors even if Python leaves the movement block early. Use speed settings of 1 to 30 and durations of at most 2 seconds; start at 15 and 0.5 seconds. Your instructor may require lower limits.

Set steering separately. The helper pauses briefly after stopping; confirm the wheels are stopped before reversing. `finish_performance()` attempts cleanup even if a previous cleanup command fails. Software cleanup does not replace physical supervision.

```python
import time
import math
from betabox_robotics import BetaboxCar


def move_for(car, direction, speed, seconds):
    if direction not in ("forward", "backward"):
        raise ValueError("Choose forward or backward.")
    if not math.isfinite(speed) or not 0 < speed <= 30:
        raise ValueError("Use a speed greater than 0 and at most 30.")
    if not math.isfinite(seconds) or not 0 < seconds <= 2:
        raise ValueError("Use a duration greater than 0 and at most 2 seconds.")

    car.stop()
    try:
        if direction == "forward":
            car.forward(speed)
        else:
            car.backward(speed)
        time.sleep(seconds)
    finally:
        car.stop()
    time.sleep(0.2)


def finish_performance(car):
    try:
        car.stop()
        car.center()
    finally:
        try:
            car.stop_audio()
        finally:
            car.close()
```
