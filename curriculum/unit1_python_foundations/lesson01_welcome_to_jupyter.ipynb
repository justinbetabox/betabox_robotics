## 3. Jupyter Remembers What You Run

Code cells can share information with each other.

Run the next cell.

```python
message = "Jupyter remembers me!"
```

You should notice that nothing appeared underneath the cell.

The code stored some information for us. We'll learn how this works in a later Python lesson.

Now run the next cell.

```python
print(message)
```

The second cell was able to use information created by the first cell.

That's because Jupyter keeps a running **kernel** behind your notebook.

The kernel runs your Python code and remembers information from code cells you have already run.

## 4. Running Cells in Order

Here's one of the most important things to remember when using Jupyter:

> **Jupyter remembers the code you have run, not simply the code you can see.**

Let's see what that means.

Run the next two cells **in order**.

```python
number = 10
```

```python
print(number)
```

You should see `10`.

Now go back to the cell that says:

    number = 10

Change it to:

    number = 20

**Do not run that cell yet.**

Instead, come back and run the `print(number)` cell again.

### What do you predict will happen?

Will Jupyter print `10` or `20`?

You should still see `10`.

Even though the code on the screen now says `number = 20`, you haven't run the changed code yet.

Now go back and run the changed `number = 20` cell.

Then run `print(number)` again.

This time you should see `20`.

> **Editing a code cell does not run it.**

When a notebook behaves differently from what you expect, check whether you have run the cells in the correct order.

## 6. Stopping Running Code

Sometimes you need to stop code before it finishes.

The next cell waits for 30 seconds.

Run it, but **don't wait for it to finish**.

While it is running, use Jupyter's **Interrupt Kernel** button or menu command to stop it.

```python
time.sleep(10)

print("Finished!")
```

After interrupting the cell, you may see a message containing:

    KeyboardInterrupt

That's okay!

It means you successfully stopped the running code.

This will be especially important when you begin writing programs that control your robot.

## 7. Starting Fresh

Sometimes the easiest way to fix a confusing notebook is to start with a fresh kernel.

Remember our `number` from earlier?

Run the next cell.

```python
print(number)
```

Now use Jupyter's **Restart Kernel** command.

You can access this command through the **Refresh** button above or from **Kernel -> Restart Kernel**.

Restarting the kernel clears everything Python was remembering.

After the restart, run the `print(number)` cell above again.

What happens?

You should see an error because the new kernel doesn't know what `number` is yet.

That's expected.

Restarting the kernel:

- stops running code
- clears information created by previous code cells
- does **not** erase the code written in your notebook

To get everything back, run the notebook's code cells again from the beginning.

## 8. Clearing Cell Outputs

Sometimes you may want to remove the results shown underneath your code cells without changing the code itself.

Jupyter lets you:

- clear the output of one code cell
- clear the outputs of every code cell in the notebook

Clearing output does **not** delete your code, and it does **not** restart the kernel.

That means Python may still remember values from code you ran earlier.

You can clear cell output by either right clicking the cell and selecting **Clear cell output** or by going  to **Edit -> Clear Cell Output** or **Edit -> Clear Outputs of All Cells**.

Try clearing the output from one of the cells above.

Then clear the outputs from all cells.

Notice that the code is still there.

### What's the Difference?

**Clear Outputs**
- removes displayed results
- keeps your code
- keeps the current kernel state

**Restart Kernel**
- keeps your code
- clears Python's remembered state
- does not necessarily clear saved outputs

**Restart Kernel and Clear Outputs**
- gives you both a fresh kernel and a clean notebook display

## 9. Run From the Beginning

Use Jupyter's **Restart Kernel and Run All Cells** command.

Watch as Jupyter runs the notebook again from top to bottom.

Running from the beginning is a great way to check that your notebook works correctly and doesn't depend on code you ran earlier by accident.
