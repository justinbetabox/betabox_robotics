## 1. Prepare Your Tools

Run the setup cell. The `read_distance()` helper requests a reading and returns `None` if the robot runtime reports an error or the number is unusable. It also displays any runtime error details. We will reuse it throughout this lab.

`None` means no usable reading, not a clear path. Do not change this helper.

```python
import math
import time
from betabox_robotics import BetaboxCar
from betabox_robotics.runtime.errors import RobotRuntimeError


def read_distance(car):
    try:
        distance = car.distance()
    except RobotRuntimeError as error:
        print("Reading failed:", error)
        return None

    if distance is None or not math.isfinite(distance) or distance <= 0:
        return None

    return distance
```

```python
car = BetaboxCar()
try:
    distance = read_distance(car)
    if distance is None:
        print("Sensor error!")
    else:
        print("Distance:", distance, "cm")
finally:
    car.close()
```
