## Try it 7: Exit a loop with break

Predict how many checks will appear. Change the target check number and run the example again.

The loop includes a maximum of 10 checks. Try a target of 12 to see what happens when the target is never reached.

````python
target = 3

for check in range(1, 11):
    print("Check:", check)

    if check == target:
        print("Target reached!")
        break

print("Checks finished.")
````

## Try it 8: Break out of a while loop

Run this example. Change the stopping value from 4 to 2 and observe which numbers appear.

Keep the counter update and the `break` condition reachable. Use a stopping value between 1 and 10.

````python
count = 0
stop_at = 4

while True:
    count = count + 1
    print("Checking:", count)

    if count >= stop_at or count >= 10:
        break

print("Done!")
````
