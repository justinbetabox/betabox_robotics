## 1. Prepare Your Tools

Run this setup cell first. The helper function stops the motors, centers the steering, and attempts to close the car even if a cleanup command fails. Each activity creates a fresh car object and uses this helper when it finishes.

Keep the cleanup code in place throughout the lesson.

```python
import time
from betabox_robotics import BetaboxCar


def finish_driving(car):
    try:
        car.stop()
        car.center()
    finally:
        car.close()
```

## 2. Try It: Your First Movement

With the driven wheels safely off the ground, run the next cell. Watch the drive wheels during the test and confirm that they stop afterward.

```python
car = BetaboxCar()

try:
    car.center()
    car.forward(20)
    time.sleep(1)
finally:
    finish_driving(car)
```
