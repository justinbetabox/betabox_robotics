## 4. Make a melody with a loop

Run this sequence, then rearrange the list. Try repeating one note and adding another. Keep your melody short.

```python
notes = ["D4", "F4", "A4", "F4"]

car = BetaboxCar()
try:
    car.stop()
    for note in notes:
        car.play_note(note, 0.3)
finally:
    finish_audio(car)
```

## 5. Compose with note-duration pairs

Each tuple specifies a note and its duration. Listen to the phrase, change `gap` from 0.1 to 0.3, then change the final note's duration. Try creating your own melody of at least five notes for your robot's startup sound.

```python
melody = [
    ("D4", 0.2),
    ("F4", 0.4),
    ("E4", 0.2),
    ("A4", 0.6),
]

gap = 0.1

car = BetaboxCar()
try:
    car.stop()
    car.play_melody(melody, gap=gap)
finally:
    finish_audio(car)
```
