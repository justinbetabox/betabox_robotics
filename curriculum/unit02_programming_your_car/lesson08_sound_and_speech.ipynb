
A file path tells Python where to find a file.


A path can also include folders:


This refers to a file inside a folder named `sounds`.

The sound file must exist at the specified path for playback to work.

## 4. Playing Musical Notes

Your Betabox car can do more than speak and play recorded sounds.

It can also generate musical notes!

The Betabox Robotics library provides a method called:

```python
car.play_note()
```

This method lets you choose a note and how long it should play.

### Understanding Pitch and Duration

Musical notes have two important properties:

- **Pitch:** How high or low a note sounds.
- **Duration:** How long the note plays.

For example:

```python
car.play_note("C4", 0.5)
```

This plays the note C4 for approximately half a second.

The first argument specifies the note, and the second specifies its duration in seconds.

### Your First Musical Notes

Let's play three different notes.

Make sure your speaker is working and the volume is comfortable.

Run the next cell to hear how the notes sound.

````python
car.play_note("C4", 0.5)
car.play_note("E4", 0.5)
car.play_note("G4", 0.5)
````

### What Happened?

Your program played three musical notes:

- C4
- E4
- G4

Each note has a different pitch.

Because we used the same duration for all three notes, each was requested to play for approximately half a second.

Both notes are C, but C5 has a higher pitch than C4.

## 5. Creating Melodies

Now that you've experimented with individual notes, let's combine them into a melody.

A **melody** is a sequence of musical notes played in a particular order.

You can create a melody by playing several notes one after another.

### Playing Notes in a Sequence

For example:

```python
car.play_note("C4", 0.5)
car.play_note("D4", 0.5)
car.play_note("E4", 0.5)
car.play_note("G4", 1.0)
```

This creates a short musical phrase.

Each note has its own pitch and duration.

### Using a List

Instead of writing a separate command for every note, we can store our notes in a list.

For example:

```python
notes = ["C4", "D4", "E4", "G4"]
```

We can then use a loop to play them:

```python
for note in notes:
    car.play_note(note, 0.5)
```

This makes it easier to change the order of the notes or add more notes to the melody.

### Using `car.play_melody()`

The Betabox Robotics library also provides a method called:

```python
car.play_melody()
```

This method lets you play a sequence of notes using a single command.

For example:

```python
car.play_melody(
    ["C4", "D4", "E4", "G4"],
    gap=0.1
)
```

The `gap` argument controls the pause between notes.

Try running the next cell to hear the melody.

````python
notes = ["C4", "D4", "E4", "G4"]

car.play_melody(notes, gap=0.1)
````

### What Happened?

Your program stored four notes in a list.

It then passed that list to `car.play_melody()`.

The method played the notes in sequence, with a short pause between them.
