## 1. Prepare Your Tools

Run this setup cell. `finish_driving()` is the cleanup helper from Lesson 2. The `drive_for()` helper combines forward, wait, and stop into one short movement. It accepts speed settings from 1 to 30 and durations greater than zero and no longer than two seconds for this project.

Use several short stages rather than increasing these limits. Do not change the helpers.

```python
import time
from betabox_robotics import BetaboxCar


def finish_driving(car):
    try:
        car.stop()
        car.center()
    finally:
        car.close()


def drive_for(car, speed, seconds):
    if not 0 < speed <= 30:
        raise ValueError("Use a speed between 1 and 30.")
    if not 0 < seconds <= 2:
        raise ValueError("Use a duration greater than 0 and at most 2 seconds.")
    try:
        car.forward(speed)
        time.sleep(seconds)
    finally:
        car.stop()
```
