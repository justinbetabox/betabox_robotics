## 3. Understanding Speed

Before we make your car move, let's talk about speed.

Your Betabox car has two motors that turn its wheels.

Python can tell those motors how quickly to spin by providing a **speed** value.

The Betabox Robotics library uses speed values from `0` to `100`.

- `0` means no motor power.
- `20` means relatively low motor power.
- `50` means medium motor power.
- `100` means maximum motor power.

These values represent motor power, not a measurement such as miles per hour.

For example:

```python
car.forward(20)
```

This tells your car to move forward using a speed value of `20`.

A higher speed value generally makes the car move faster, but the actual speed depends on factors such as battery voltage, the surface, and the motors.

### Start Slowly

When learning to drive your car, always begin with a low speed.

We'll use `20` for our first movement test.

You can experiment with other speeds later.

## 4. Moving Forward

Now it's time to make your car move!

Your Betabox car provides a method named:

```python
car.forward(speed)
```

The `forward()` method tells the motors to turn in the direction that moves the car forward.

The number inside the parentheses is the speed value.

For example:

```python
car.forward(20)
```

This tells the car to move forward at a low speed.

### Your First Movement Test

Make sure your car is securely supported with its driven wheels off the ground.

We'll make the wheels spin for **one second**, then stop them.

We use `time.sleep(1)` to tell Python to wait for one second.

The `try` and `finally` keywords help ensure the stop command runs even if an ordinary Python error or interruption occurs.

You don't need to memorize those keywords yet. For now, focus on what the movement commands do.

Run the next cell.

````python
try:
    car.forward(20)
    time.sleep(1)
finally:
    car.stop()
````

### What Happened?

Your car should have:

1. Started turning its wheels forward.
2. Continued for approximately one second.
3. Stopped its motors.

Look at the first instruction:

```python
car.forward(20)
```

This is a **method call**, just like the ones you learned about in Lesson 1.

However, there's an important difference.

Previously, you used methods such as `car.distance()` to **receive information** from the car.

Now you're using `car.forward()` to **send instructions** to the car.

### Why Did We Use `time.sleep()`?

Without a delay, Python would immediately move to the next instruction.

By using:

```python
time.sleep(1)
```

we tell Python to wait for one second before continuing.

During that time, the motors continue running.

### Why Did We Stop the Car?

The `forward()` method does not automatically stop the motors after a certain amount of time.

That's why we call:

```python
car.stop()
```

The stop command tells the motors to stop turning.

**Remember:** Always include a safe way to stop your car when writing movement programs.

## 5. Stopping Your Car

You've already used one of the most important methods in the Betabox Robotics library:

```python
car.stop()
```

The `stop()` method tells your car's drive motors to stop turning.

Unlike `forward()`, it doesn't need a speed argument.

### Moving and Stopping Are Separate Instructions

Consider these two instructions:

```python
car.forward(20)
car.stop()
```

The first instruction tells the car to move forward.

The second tells it to stop.

Python executes these instructions one after another, so the motors would receive the stop command almost immediately.

To make the car move for a noticeable amount of time, we need to wait between those instructions.

### Stopping Versus Closing

Remember that `stop()` and `close()` do different jobs.

**`car.stop()`**
- Stops the drive motors.
- Keeps your car object active.
- Allows you to issue more driving commands afterward.

**`car.close()`**
- Stops active operations and releases hardware control.
- Closes the car object.
- Requires you to create a new car object before using the car again.

While driving, you'll use `car.stop()` frequently.

When you're completely finished, you'll use `car.close()`.

### Try It

Make sure your car's wheels are no longer moving.

Run the next cell.

````python
car.stop()
````

Nothing may appear to happen.

That's okay!

Your car was already stopped, so the command simply told it to remain stopped.

You can call `car.stop()` even when the motors aren't moving, as long as the car object is still active.

**Remember:** Stopping the motors does not close your car.

## 6. Controlling Movement Duration

Now that you know how to start and stop the motors, let's control how long they run.

In Python, the `time.sleep()` function pauses your program for a specified number of seconds.

For example:

```python
time.sleep(1)
```

pauses Python for approximately one second.

You can also use decimal values:

```python
time.sleep(0.5)
```

This pauses Python for approximately half a second.

### Combining Movement and Time

By combining three instructions, we can control how long the car moves:

1. Start the motors.
2. Wait for a specified amount of time.
3. Stop the motors.

We'll use `try` and `finally` again to make sure the stop instruction runs when the code exits normally or encounters an ordinary Python exception.

### Try a Two-Second Movement

Make sure your car is securely supported with its driven wheels off the ground.

Run the next cell.

````python
try:
    car.forward(20)
    time.sleep(2)
finally:
    car.stop()
````

### What Happened?

Your car should have moved its wheels forward for approximately two seconds before stopping.

Compare that with the one-second movement from Section 4.

The speed stayed the same, but the **duration** changed.

This is an important distinction:

- **Speed** controls how quickly the motors turn.
- **Duration** controls how long the motors continue turning.

## 8. Understanding Steering

Your Betabox car has two different systems that work together to control how it moves.

**Drive motors** make the wheels turn to move the car forward or backward.

**Steering** changes the direction the front wheels point.

These are two separate controls.

For example, a real car can have its front wheels turned to the left even while it is parked.

Your Betabox works in a similar way.

### Moving Versus Steering

Think about these two instructions:

```python
car.forward(20)
```

This tells the drive motors to turn.

Now consider:

```python
car.left(20)
```

This tells the steering system to turn the front wheels to the left.

The important difference is:

- `forward()` controls movement.
- `left()` controls steering.

**Turning the steering wheels does not automatically make the car drive forward.**

The car needs both steering and movement to follow a curved path.

### Steering Angles

Your Betabox uses an **angle** to describe how far the front wheels should turn.

Angles are measured in degrees.

For example:

```python
car.left(15)
```

requests a 15-degree steering angle to the left.

A larger angle generally produces a sharper turn when the car is moving.

Your Betabox's steering range is limited to protect the mechanism, so you should only use supported angles.

We'll begin with small steering angles and learn how to return the wheels to the center.


The `angle` argument controls how far the wheels turn.

For example:

```python
car.left(15)
```

turns the steering to the left by a commanded angle of 15 degrees.

Similarly:

```python
car.right(15)
```

turns the steering to the right by a commanded angle of 15 degrees.

The number `15` represents the requested steering angle.

## 10. Centering the Wheels

So far, you've learned how to steer left and right.

But what if you want the car to travel straight again?

Your Betabox car provides another method:

```python
car.center()
```

The `center()` method returns the steering system to its calibrated center position.

When the front wheels are centered, the car is set up to travel approximately straight ahead.

### Try Centering the Wheels

Your wheels should still be pointing to the left or right from the previous exercise.

Run the next cell:

````python
car.center()
````

### What Happened?

The front wheels should have returned to their center position.

Notice that we didn't need to provide an angle.

The `center()` method already knows the steering system's calibrated center position.

### Steering Doesn't Automatically Reset

Consider these instructions:

```python
car.left(20)
car.forward(20)
```

The first instruction turns the steering wheels left.

The second starts the drive motors.

The car would follow a curved path because the steering is still pointing left.

Stopping the drive motors does not automatically center the steering.

For example:

```python
car.stop()
```

stops the motors, but the steering remains in its previous position.

To return the steering to the center, you need:

```python
car.center()
```

This is an important habit when writing driving programs.

## 11. Combining Movement and Steering

You've learned how to control your car's drive motors and steering separately.

Now let's combine them!

When your car moves forward with its steering centered, it should travel approximately straight ahead.

When the front wheels are turned left or right, the car follows a curved path.

We can create these movements by combining methods you've already learned.

### Steering Before Moving

Consider this program:

```python
car.left(20)

```

The first instruction turns the front wheels to the left.

The second instruction starts the drive motors.

Because the wheels are pointing left, the car would follow a curved path instead of traveling straight ahead.

However, we still need to control how long the car moves and make sure it stops safely.

### What Happened?

Your car should have:

Instead, you observed how steering and movement work together.

### Why Did We Center the Steering?

Notice that our program ends with two methods:

```python
car.stop()
car.center()
```

These instructions perform different jobs.

- `stop()` stops the drive motors.
- `center()` returns the steering to its center position.

Using both commands leaves the car ready for its next movement.

## 12. Stopping and Closing

You've learned how to move your car, steer its wheels, and stop its motors.

Before finishing, let's review two important methods:

```python
car.stop()
car.close()
```

These methods have different purposes.

### Stopping Your Car

The `stop()` method stops the drive motors.

```python
car.stop()
```

After stopping, your car object is still active.

You can continue using it to:

- move forward or backward
- turn the steering left or right
- read sensor measurements
- perform other supported actions

Stopping is something you'll do frequently while driving.

### Closing Your Car

The `close()` method finishes your session with the car.

```python
car.close()
```

Closing the car:

- stops active operations
- releases any hardware control held by your program
- closes the car object

Once the car is closed, you cannot continue using that same object.

To use the car again, you'll need to create a new one:

```python
car = BetaboxCar()
```

### Remember the Difference

**Stop when you want to pause movement.**

**Close when you're completely finished using the car.**

### Why Does This Matter?

Your Betabox can be used by different applications, including Jupyter notebooks and Manual Drive.

Only one application can control the motors and servos at a time.

Closing your car releases any control held by your notebook so another application can use the hardware.

This helps prevent different programs from trying to move the car at the same time.

````python
try:
    car.stop()
    car.center()
finally:
    car.close()
````

### Your Car Is Now Closed

Your program has finished working with the car.

The drive motors should be stopped, and the car object is closed.

If you want to use your car again, you'll need to create a new object.

Remember the basic lifecycle you learned in Lesson 1:

**Create → Use → Close**

This pattern will continue to be important as your programs become more advanced.

### A Good Driving Habit

Whenever you write a driving program:

- Start with a safe speed.
- Use short, controlled movements.
- Stop the motors when movement is finished.
- Center the steering when appropriate.
- Close the car when you're completely finished.

These habits will help you write safer and more predictable robotics programs.
