## 1. How Grayscale Sensors Work

Your Betabox car has three grayscale sensors mounted underneath its chassis.

These sensors help your robot detect differences between light and dark surfaces.

### Measuring Reflected Light

Grayscale sensors work by detecting how much light reflects off a surface.

Different surfaces reflect different amounts of light.

For example:

- **White surfaces** generally reflect more light.
- **Black surfaces** generally reflect less light.
- **Gray surfaces** reflect an amount of light somewhere between the two.

By measuring these differences, your robot can recognize changes in the surface beneath it.

### Three Sensors Working Together

Your Betabox has three grayscale sensors:

- **Left sensor:** Measures the surface beneath the left side.
- **Center sensor:** Measures the surface beneath the middle.
- **Right sensor:** Measures the surface beneath the right side.

Each sensor produces its own measurement.

By comparing all three readings, your program can determine which parts of the sensor array are positioned over a dark or light surface.

### Why Use Three Sensors?

Imagine placing a strip of black tape on a white surface.

If the center sensor detects the black tape while the left and right sensors detect white, the line is approximately centered beneath the sensor array.

If the left sensor detects black while the center and right sensors detect white, the line may be positioned toward the left side of the car.

These differences will become especially useful when we build a line-following robot in Lesson 7.

### What Can Affect the Measurements?

Grayscale sensors can be affected by:

- ambient lighting
- surface color and material
- reflections from shiny surfaces
- the distance between the sensors and the ground
- shadows and uneven surfaces

Because of these differences, a measurement that represents black on one surface might not represent black on another.

We'll learn how to choose a threshold that works for our testing environment.

## 3. Understanding Thresholds

In the previous section, you measured the values returned by your grayscale sensors over light and dark surfaces.

You probably noticed that the readings changed depending on the surface beneath the car.

But how can Python decide whether a sensor is detecting a light or dark surface?

We can use something called a **threshold**.

### What Is a Threshold?

A threshold is a value used to separate measurements into different categories.

For example, imagine a sensor produces these readings:

| Surface | Measurement |
|---|---|
| White paper | 850 |
| Black tape | 250 |

These are example values. Your actual readings may be different.

We could choose a threshold of `550`, halfway between the two measurements.

```python
threshold = 550
```

If this sensor produces higher readings over white surfaces, we could classify its measurements using:

```python
if reading < threshold:
    print("Dark surface detected!")
else:
    print("Light surface detected!")
```

This works because the example sensor produces lower readings over dark surfaces.

However, not every sensor setup behaves the same way.

**Always compare your actual measurements before deciding which values represent dark or light surfaces.**

### Choosing Your Own Threshold

Look back at the measurements you recorded in Section 2.

Choose one sensor and compare its readings over white paper and black tape.

A simple starting threshold is halfway between the two measurements.

For example:

```python
white_value = 850
black_value = 250

threshold = (white_value + black_value) / 2

print("Threshold:", threshold)
```

This produces:

```text
Threshold: 550.0
```

Your threshold should be based on your own sensor readings, not these example values.

### Why Does Calibration Matter?

Different surfaces, lighting conditions, and sensor positions can produce different measurements.

A threshold that works well on one surface might not work as well on another.

Choosing a threshold based on actual measurements is a simple form of **calibration**.

Calibration helps your program interpret sensor measurements more reliably.

Instead of working with raw measurements, you can use the results to make decisions about the surface beneath each sensor.

### Raw Measurements vs. Sensor Status

You've already used:


This returns the raw measurements from the left, center, and right sensors.

The `line_status()` method provides a simpler way to interpret those readings.

You can also supply a threshold:

The threshold should be chosen using your sensor measurements and calibration.
